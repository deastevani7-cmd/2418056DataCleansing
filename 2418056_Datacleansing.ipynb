3. Membuat Fungsi Pembersih

```python
SINGKATAN = {"IT", "HRD"}
FORMAT_TANGGAL = ["%Y-%m-%d", "%Y/%m/%d",
                  "%d-%m-%Y", "%d/%m/%Y"]


def bersihkan_teks(x):
    """Hapus spasi di awal/akhir dan spasi ganda di tengah."""
    return " ".join(str(x).split())


def rapikan_departemen(x):
    x = bersihkan_teks(x)
    return x.upper() if x.upper() in SINGKATAN else x.title()


def ubah_tanggal(x):
    """Semua format tanggal -> YYYY-MM-DD."""
    x = bersihkan_teks(x)
    for fmt in FORMAT_TANGGAL:
        try:
            return pd.to_datetime(x, format=fmt)
        except ValueError:
            continue
    return pd.NaT
```

4. Menghapus Spasi Berlebih

```python
for k in kolom:
    df[k] = df[k].apply(bersihkan_teks)

df.head()
```

5. Menyeragamkan Huruf dan Tipe Data

```python
df["ID_Karyawan"]   = df["ID_Karyawan"].astype(int)
df["Nama"]          = df["Nama"].str.title()
df["Jenis_Kelamin"] = df["Jenis_Kelamin"].str.upper()
df["Departemen"]    = df["Departemen"].apply(rapikan_departemen)
df["Jabatan"]       = (df["Jabatan"].str.title()
                       .str.replace(r"\bHr\b", "HR", regex=True))
df["Kota"]          = df["Kota"].str.title()
df["Status"]        = df["Status"].str.title()

df.head()
```

6. Menyeragamkan Format Tanggal

```python
df["Tanggal_Masuk"] = df["Tanggal_Masuk"].apply(ubah_tanggal)

print("Tanggal gagal dikonversi:", df["Tanggal_Masuk"].isna().sum())
df["Tanggal_Masuk"] = df["Tanggal_Masuk"].dt.strftime("%Y-%m-%d")
df.head(10)
```

7. Menghapus Duplikat dan Mengurutkan Data

```python
df = df.drop_duplicates(subset="ID_Karyawan").sort_values("ID_Karyawan")
df = df.reset_index(drop=True)
print("Jumlah baris:", len(df))
```
